#Data leakage
Data leakage in machine learning happens when information from outside the training dataset is improperly used to build a model. This causes the model to gain access to data it would not have during real-world predictions.

As a result, the model shows artificially high accuracy during training and validation, but fails or behaves poorly once deployed in production

```python
leakage_columns=['rejection_reason',
                 'claim_settlement_type',
                 'settlement_ratio_pct',
                 'settled_amount_inr']

df.drop(leakage_columns,axis=1,inplace=True)
```

#Divide the dataset into independent  variables and dependent variable

```python
x=df.drop('claim_status',axis=1)
y=df['claim_status']
```

#Identify numerial,boolean and Categorical columns

```python
categorical_columns = []
numerical_columns = []
boolean_columns = []
for column in x.columns:
  if x[column].dtype == 'object':
    categorical_columns.append(column)
  elif x[column].dtype == 'bool':
    boolean_columns.append(column)
  else:
    numerical_columns.append(column)
```

```python
categorical_columns
```

#Label Encoding

```python
!pip install category_encoders
```

```python
import category_encoders as ce
encoder = ce.OrdinalEncoder(cols=categorical_columns)
x[categorical_columns] = encoder.fit_transform(x[categorical_columns])

x.head()
```

#Feature Scaling using MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x[numerical_columns] = scaler.fit_transform(x[numerical_columns])
x.head()
```

#Split the dataset into Training & Testing data


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

#Model Training

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

Using GridSearch CV
